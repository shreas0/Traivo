```python
import pandas as pd

df = pd.read_csv(r"C:\Users\shres\Documents\Travel Planner\backend\data_processing\traivo_travel_customer_reviews.csv")

print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

df.head()
```

```python
DESTINATION_COL = "destination"
REVIEW_COL = "review_text"
SENTIMENT_COL = "sentiment"

required = [DESTINATION_COL, REVIEW_COL, SENTIMENT_COL]
```

```python
df[DESTINATION_COL] = df[DESTINATION_COL].astype(str).str.strip()
df[REVIEW_COL] = df[REVIEW_COL].astype(str).str.strip()
df["_sentiment"] = (
    df[SENTIMENT_COL].astype(str).str.strip().str.lower()
)
df = df[
    (df[DESTINATION_COL] != "") &
    (df[REVIEW_COL] != "")
].copy()
```

```python
positive_df=df[df["sentiment"]=="positive"]
```

```python
print("Total reviews:", len(df))
print("Positive reviews:", len(positive_df))
print("Unique destinations:", positive_df[DESTINATION_COL].nunique())
```

```python
def search_reviews(destination):
    reviews = df[
        (df["destination"].str.lower() == destination.lower()) &
        (df["sentiment"] == "positive")
    ]["review_text"]

    return "\n".join(reviews)
```

```python
import pickle

positive_df = df[
    df["sentiment"].str.lower() == "positive"
][["destination", "review_text"]].dropna().copy()

with open("positive_reviews.pkl", "wb") as file:
    pickle.dump(positive_df, file)

print("Search data exported!")
```
